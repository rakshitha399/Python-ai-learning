# using while loop

```python
l = [1,2,3,4,5]
```

# skip by index using i ==2

```python
# USING INDEXING 
i = 0
while i<len(l):
    if i==2: #index skip at index 2 == 3
        i=i+1
        continue
    print(l[i])
    i = i+1
```

# skip by value using l[i]

```python
i = 0
while i<len(l):
    if l[i]==2: # value
        i=i+1
        continue
    print(l[i])
    i = i+1
print(len(l))
```

# Factorial

```python
fact = 1

for num in range(1,n+1):
    fact = fact*num
    print(num,":",fact)
print("Sum of factorail :",fact)
```

# nesting loop 
  # useful in pattern questions

```python
for i in  range(1,5):
    print(i)
    for j in range(i):
        print(j)
    print()
```

```python
for var1 in range(1,11): # 1 to 10
    if var1==5:
        break
    print("outer loop n0:", var1)
    for var2 in range(1,5): # 3 to 4
        print("inner loop :",var1*var2)
    print("outside the inner loop for",var1)
    
```

```python
for i in range(1,11):
    print("*")
    for j in range(1,4):
        print("*i",end = "")
    print("*")
```
