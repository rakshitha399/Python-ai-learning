# list comprehension

#its a faster and easy way to create list using a single line

#syntax : [element for element in iterable if condition] 

```python
l = [ ]
for i in range(1,50+1):
    l.append(i)
print(l,end="")
```

```python
l = [i for i in range(1,50)]
l
```

```python
num = [f"even:{i}" if i%2==0 else f"odd:{i}" for i in range(0,10+1)]
num
```

#anther way is nested list comprehension

```python
num1 = [[i , "odd" if i%2!=0 else "even"]for i in  range(10)]
num1
```

#just if 

```python
if_only = [[i ,"even"] for i in range(11) if i%2==0]
if_only
```

```python
com = [ch if ch not in  "aeiou" else "vowels" for ch in "tulsi"]
com
```

```python
[[i for i in range(num) ] for num in range(1,5)]
```
